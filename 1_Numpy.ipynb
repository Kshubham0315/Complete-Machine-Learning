```python
#This is Z-score normalization, commonly used before training algorithms like Logistic Regression, SVM, and Neural Networks.

X = np.array([10, 20, 30, 40, 50])

X_scaled = (X - np.mean(X)) / np.std(X)

X_scaled
```
