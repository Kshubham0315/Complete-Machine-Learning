# Interview Questions ⭐⭐⭐⭐⭐
# Q1. What is a histogram?

A histogram visualizes the distribution of continuous numerical data by grouping values into intervals (bins).

Q2. What are bins?

Bins are intervals used to group numerical values before counting their frequencies.

Q3. Difference between Histogram and Bar Chart?
Histogram	Bar Chart
Continuous Data	Categorical Data
Shows Distribution	Shows Comparison
Bars Touch	Bars Have Gaps
Q4. Why use density=True?

To normalize the histogram so the area under the histogram equals 1, making it suitable for comparing distributions and probability density.

Q5. What does plt.hist() return?

A tuple containing:

counts, bins, patches
counts → Frequency in each bin
bins → Bin edges
patches → Rectangle objects used to draw the bars
🎯 Complete Histogram Cheat Sheet
Parameter	Purpose
bins	Number of intervals
color	Fill color
edgecolor	Border color
alpha	Transparency
density	Normalize frequencies
cumulative	Cumulative counts
range	Restrict displayed values
orientation	Vertical/Horizontal
stacked	Stack multiple histograms
histtype	Bar, step, etc.
weights	Weighted histogram
log	Logarithmic y-axis

# Module 6 — Scatter Plot & Correlation Analysis

````python
import matplotlib.pyplot as plt

height = [160,165,170,175,180]

weight = [55,58,65,72,80]

plt.figure(figsize=(6,6))

# plt.scatter(height , weight)
# s- size of the marker
# c- color of the marker
# alpha- transparency of the marker

colors = [
    "red",
    "green",
    "blue",
    "orange",
    "purple"
]

population = [
    300,
    500,
    900,
    1200,
    2000
]

scatter = plt.scatter(height , weight , color=colors , s=population , alpha=0.5 , marker="^",edgecolors="black" , linewidth=2)
plt.colorbar(scatter)

plt.xlabel("Height")

plt.ylabel("Weight")

plt.title("Height vs Weight")

# plt.grid(True)

plt.show()
````

````python
salary = [
    30000,
    40000,
    60000,
    80000,
    100000
]

plt.scatter(
    height,
    weight,
    c=salary,
    cmap="viridis"
)
````

## 💡 Section 1: Interview Q&A

> **Q1. What core information does a Box Plot display?**
> 
> A box plot visualizes the **Five-Number Summary** along with anomalies:
> * 📉 **Minimum** (Lower whisker)
> * 🔲 **Q1** (First Quartile / 25th Percentile)
> * ➖ **Median** (50th Percentile)
> * 🔲 **Q3** (Third Quartile / 75th Percentile)
> * 📈 **Maximum** (Upper whisker)
> * 🔴 **Outliers** (Individual points beyond whiskers)

---

> **Q2. What is IQR?**
> 
> The **Interquartile Range (IQR)** measures statistical dispersion (spread of the middle 50% of data):
> ```python
> IQR = Q3 - Q1
> ```

---

> **Q3. How are outliers programmatically detected?**
> 
> Using the standard **$1.5 \times \text{IQR}$ rule**:
> ```python
> lower_bound = Q1 - (1.5 * IQR)
> upper_bound = Q3 + (1.5 * IQR)
> 
> # Any value < lower_bound or > upper_bound is an outlier!
> ```

## 💡 Section 1: Interview Q&A

> **Q1. What is the difference between `subplot()` and `subplots()`?**
> 
> * `subplot()` creates or selects one subplot at a time (legacy approach).
> * `subplots()` creates a `Figure` and one or more `Axes` together all at once, making it the **preferred modern API**.

---

> **Q2. What does `sharex=True` do?**
> 
> It forces all selected subplots to **share the same X-axis**, significantly improving visual alignment, scaling consistency, and clean comparisons.

---

> **Q3. Why use `GridSpec`?**
> 
> To create **custom, complex layouts** where subplots have different custom sizes or span across multiple rows and columns.

---

> **Q4. Why use `twinx()`?**
> 
> To plot **two datasets with completely different Y-axis scales** sharing the exact same X-axis on a single plot area.

---

> **Q5. Why flatten Axes?**
> 
> When dealing with multi-dimensional subplot grids, `.flatten()` converts the Axes array into 1D, which **simplifies iteration over multiple subplots** inside loops.

# 11. Common Mistakes

❌ Using `barplot()` to count observations

Use

```python
sns.countplot()
```

---

❌ Forgetting `annot=True`

Wrong

```python
sns.heatmap(corr)
```

Correct

```python
sns.heatmap(
    corr,
    annot=True
)
```

---

❌ Ignoring Outliers

Always check

```python
sns.boxplot()
```

before training a model.

---

❌ Ignoring Class Imbalance

Always visualize

```python
sns.countplot(
    data=df,
    x="target"
)
```

# 12. Interview Questions

### Q1. Difference between countplot() and barplot()?

| countplot() | barplot() |
|-------------|-----------|
| Counts observations | Aggregates numerical values (mean by default) |

---

### Q2. Difference between scatterplot() and regplot()?

| scatterplot() | regplot() |
|---------------|-----------|
| Scatter only | Scatter + Regression Line |

---

### Q3. Difference between boxplot() and violinplot()?

| Box Plot | Violin Plot |
|----------|-------------|
| Statistical summary | Summary + Distribution |

---

### Q4. Difference between histplot() and kdeplot()?

| histplot() | kdeplot() |
|------------|-----------|
| Bars | Smooth Density Curve |

---

### Q5. Why use heatmap()?

To visualize the **Correlation Matrix**, helping identify highly correlated features and supporting feature selection.

# 🎯 Conclusion

Seaborn is **not a replacement for Matplotlib**—it is a higher-level library built on top of it.

For AI/ML projects:

- **Matplotlib** → Fine-grained customization and publication-quality figures.
- **Seaborn** → Fast, elegant statistical visualizations and EDA.

Master these core plots, and you'll be well-prepared for most AI/ML interviews and real-world data analysis tasks.
