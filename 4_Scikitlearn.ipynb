# ⭐ Interview Questions

## Q1. What is Data Preprocessing?

**Answer:**

Data preprocessing is the process of cleaning, transforming, and organizing raw data into a format that is suitable for machine learning models.

---

## Q2. Why is Data Preprocessing Important?

**Answer:**

Machine learning models cannot learn effectively from raw or inconsistent data.

Preprocessing helps by:

- Removing missing values
- Handling duplicate records
- Correcting incorrect data types
- Preparing features for model training
- Improving model accuracy and performance

---

## Q3. How do you check missing values?

```python
df.isnull().sum()
```

---

## Q4. How do you remove duplicate rows?

```python
df.drop_duplicates()
```

---

## Q5. What is the difference between Features and Target?

| Features (X) | Target (y) |
|--------------|------------|
| Input variables | Output variable |
| Independent variables | Dependent variable |
| Used to make predictions | Value to be predicted |

---

## Q6. What is Data Leakage?

**Answer:**

Data leakage occurs when information from the test set (or future data) is accidentally used during model training.

As a result, the model appears to perform extremely well during evaluation but fails on real-world unseen data.

# Normalizer

````python
# StandardScaler → Column-wise scaling
# Normalizer → Row-wise scaling
````

Why do we use Normalizer?

Suppose you have document vectors:

Math	AI	Python
3	4	0
6	8	0

Row 1:

[3, 4, 0]

Row 2:

[6, 8, 0]

Notice:

Row 2 is just twice Row 1.

Many ML algorithms (like KNN, Cosine Similarity, and Text Classification) care about the direction of the vector, not its size.

Normalizer makes both rows have the same length while preserving their direction.

````python
from sklearn.preprocessing import Normalizer

X = [
    [3, 4],
    [6, 8]
]

norm = Normalizer()

X_norm = norm.fit_transform(X)

X_norm
````

    StandardScaler vs Normalizer
    StandardScaler	Normalizer
    Works column-wise	Works row-wise
    Uses mean & standard deviation	Uses vector length (norm)
    Centers data around 0	Makes each sample length = 1
    Common in Regression, SVM, Neural Networks	Common in KNN, Text Mining, Cosine Similarity

    When to Use Normalizer?

    Use it when:

    Text Classification (TF-IDF vectors)
    Cosine Similarity
    K-Nearest Neighbors (KNN)
    Recommendation Systems
    Information Retrieval

    Avoid it when:

    Linear Regression
    Logistic Regression
    Decision Trees
    Random Forest

    These algorithms usually benefit more from StandardScaler or MinMaxScaler.

Interview Questions
Q1. What is the difference between StandardScaler and Normalizer?

Answer:

StandardScaler scales features (columns) using mean and standard deviation.
Normalizer scales samples (rows) so that each sample has a unit norm (length = 1).
Q2. Does Normalizer calculate mean and standard deviation?

Answer: No. It only divides each row by its chosen norm (L1, L2, or Max).

Q3. What are the types of Normalizer?
from sklearn.preprocessing import Normalizer

# Default
Normalizer(norm="l2")

# L1 Norm
Normalizer(norm="l1")

# Max Norm
Normalizer(norm="max")
L2 (default): Divide by Euclidean length.
L1: Divide by the sum of absolute values.
Max: Divide by the largest absolute value in the row.

    If an interviewer asks:

    "How do you preprocess a dataset with both numerical and categorical features?"

    A strong answer is:

    Split the data into training and testing sets.
    Apply a scaler (such as StandardScaler) to numerical features.
    Apply OneHotEncoder to nominal categorical features.
    Apply OrdinalEncoder to ordinal categorical features if needed.
    Use ColumnTransformer to combine these preprocessing steps.
    Fit the preprocessing on the training data and transform both training and test data.
    Train the model using the transformed features.

    This is the preprocessing workflow you'll see in most production-grade Scikit-learn projects.

    Interview Questions ⭐⭐⭐⭐⭐
    Q1. What is a Linear Model?

    A model that assumes a linear relationship between features and the target.

    Q2. When should you use Linear Regression?

    When the target variable is continuous, such as predicting price, salary, or temperature.

    Q3. When should you use Logistic Regression?

    When the target variable is categorical, such as Yes/No, Spam/Not Spam, or Fraud/Not Fraud.

    Q4. What does model.coef_ return?

    It returns the learned coefficients (weights) for each feature.

    Q5. What does model.intercept_ return?

    It returns the intercept (bias) learned during training.

    Q6. Why is Logistic Regression called "Regression" even though it performs classification?

    Because it models a linear combination of features and then applies the sigmoid function to estimate probabilities, which are converted into class labels.

    Q7. What is the difference between predict() and predict_proba()?
    predict()	predict_proba()
    Returns predicted class labels	Returns probabilities for each class

    Example

    model.predict(X_test)

    [1, 0, 1]
    model.predict_proba(X_test)

    [[0.05, 0.95],

    🧠 AI/ML Interview Tip

    If an interviewer asks:

    "How do you decide whether to use Linear Regression or Logistic Regression?"

    A strong answer is:

    If the target variable is continuous (price, salary, temperature), use Linear Regression.
    If the target variable is categorical (spam/not spam, disease/no disease, fraud/not fraud), use Logistic Regression.
    Although Logistic Regression has "regression" in its name, it is fundamentally a classification algorithm because it predicts probabilities and assigns class labels.

Interview Questions ⭐⭐⭐⭐⭐
Q1. What is Classification?

Classification is a supervised learning task where the model predicts discrete categories or class labels.

Q2. Difference between Regression and Classification?
Regression	Classification
Predicts continuous values	Predicts categories
Price Prediction	Spam Detection
Q3. Why does KNN require scaling?

Because KNN uses distance calculations. Features with larger numerical values can dominate the distance if data is not scaled.

Q4. Why doesn't Random Forest require scaling?

Because Decision Trees split data based on feature thresholds, not distances.

Q5. What is overfitting in Decision Trees?

When the tree becomes too complex and memorizes the training data instead of learning general patterns.

Q6. Why is Random Forest generally better than a single Decision Tree?

Random Forest combines predictions from many trees, reducing variance and improving generalization.

Q7. Which algorithm is best for text classification?

Naive Bayes is a strong baseline because it is fast and performs well on many text classification tasks.
