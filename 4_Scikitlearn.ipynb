# ⭐ Interview Questions

## Q1. What is Data Preprocessing?

**Answer:**

Data preprocessing is the process of cleaning, transforming, and organizing raw data into a format that is suitable for machine learning models.

---

## Q2. Why is Data Preprocessing Important?

**Answer:**

Machine learning models cannot learn effectively from raw or inconsistent data.

Preprocessing helps by:

- Removing missing values
- Handling duplicate records
- Correcting incorrect data types
- Preparing features for model training
- Improving model accuracy and performance

---

## Q3. How do you check missing values?

```python
df.isnull().sum()
```

---

## Q4. How do you remove duplicate rows?

```python
df.drop_duplicates()
```

---

## Q5. What is the difference between Features and Target?

| Features (X) | Target (y) |
|--------------|------------|
| Input variables | Output variable |
| Independent variables | Dependent variable |
| Used to make predictions | Value to be predicted |

---

## Q6. What is Data Leakage?

**Answer:**

Data leakage occurs when information from the test set (or future data) is accidentally used during model training.

As a result, the model appears to perform extremely well during evaluation but fails on real-world unseen data.

# Normalizer

````python
# StandardScaler → Column-wise scaling
# Normalizer → Row-wise scaling
````

Why do we use Normalizer?

Suppose you have document vectors:

Math	AI	Python
3	4	0
6	8	0

Row 1:

[3, 4, 0]

Row 2:

[6, 8, 0]

Notice:

Row 2 is just twice Row 1.

Many ML algorithms (like KNN, Cosine Similarity, and Text Classification) care about the direction of the vector, not its size.

Normalizer makes both rows have the same length while preserving their direction.

````python
from sklearn.preprocessing import Normalizer

X = [
    [3, 4],
    [6, 8]
]

norm = Normalizer()

X_norm = norm.fit_transform(X)

X_norm
````

    StandardScaler vs Normalizer
    StandardScaler	Normalizer
    Works column-wise	Works row-wise
    Uses mean & standard deviation	Uses vector length (norm)
    Centers data around 0	Makes each sample length = 1
    Common in Regression, SVM, Neural Networks	Common in KNN, Text Mining, Cosine Similarity

    When to Use Normalizer?

    Use it when:

    Text Classification (TF-IDF vectors)
    Cosine Similarity
    K-Nearest Neighbors (KNN)
    Recommendation Systems
    Information Retrieval

    Avoid it when:

    Linear Regression
    Logistic Regression
    Decision Trees
    Random Forest

    These algorithms usually benefit more from StandardScaler or MinMaxScaler.

Interview Questions
Q1. What is the difference between StandardScaler and Normalizer?

Answer:

StandardScaler scales features (columns) using mean and standard deviation.
Normalizer scales samples (rows) so that each sample has a unit norm (length = 1).
Q2. Does Normalizer calculate mean and standard deviation?

Answer: No. It only divides each row by its chosen norm (L1, L2, or Max).

Q3. What are the types of Normalizer?
from sklearn.preprocessing import Normalizer

# Max Norm
Normalizer(norm="max")
L2 (default): Divide by Euclidean length.
L1: Divide by the sum of absolute values.
Max: Divide by the largest absolute value in the row.

    If an interviewer asks:

    "How do you preprocess a dataset with both numerical and categorical features?"

    A strong answer is:

    Split the data into training and testing sets.
    Apply a scaler (such as StandardScaler) to numerical features.
    Apply OneHotEncoder to nominal categorical features.
    Apply OrdinalEncoder to ordinal categorical features if needed.
    Use ColumnTransformer to combine these preprocessing steps.
    Fit the preprocessing on the training data and transform both training and test data.
    Train the model using the transformed features.

    This is the preprocessing workflow you'll see in most production-grade Scikit-learn projects.

    If an interviewer asks:

    "How do you decide whether to use Linear Regression or Logistic Regression?"

    A strong answer is:

````python
# Q1. What is Clustering?

# Clustering is an unsupervised learning technique that groups similar data points together without using labels.

# Q2. Why is K-Means called "K-Means"?

# Because it creates K clusters, and each cluster is represented by the mean (centroid) of its points.

# Q3. What is the Elbow Method?

# It is a technique for selecting the optimal number of clusters by plotting WCSS against different values of K and choosing the point where the curve bends.

# Q4. What is WCSS?

# Within Cluster Sum of Squares measures how close data points are to their assigned cluster centroid.

# Q5. What is the Silhouette Score?

# A metric that measures how well-separated clusters are. It ranges from -1 to +1, with values closer to +1 indicating better clustering.

# Q6. Difference between K-Means and DBSCAN?
# K-Means	DBSCAN
# Requires K	No K needed
# Sensitive to outliers	Detects outliers
# Best for spherical clusters	Works for irregular clusters
# Q7. Why should we scale data before K-Means?

# Because K-Means relies on distance calculations. Features with larger scales can dominate the clustering if data is not standardized.
````

Q5. Difference between MAE and MSE?

MAE treats all errors equally.

MSE penalizes larger errors more heavily.

Q6. What is R² Score?

It measures how much of the variance in the target variable is explained by the model.

Q1. What is a hyperparameter?

A hyperparameter is a value set before training that controls the learning process (e.g., max_depth, n_estimators, C, k).

Q2. Difference between Grid Search and Random Search?
Grid Search	Random Search
Tries every combination	Tries random combinations
More accurate	Faster
Slower	More efficient for large search spaces
Q3. What is Cross Validation?

Cross Validation repeatedly splits the data into different training and validation sets to produce a more reliable estimate of model performance.
