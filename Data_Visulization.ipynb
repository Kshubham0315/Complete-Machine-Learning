# 5. Basic Line Plot

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,25,30,35]

plt.plot(x, y)

plt.show()
```

Use Case:

* Sales Trend
* Stock Price
* Temperature
* Model Loss

---

# 6. Line Plot

```python
plt.plot(
    x,
    y,
    marker="o",
    linestyle="--",
    linewidth=2,
    color="blue"
)

plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Best Used For

* Time Series
* Trend Analysis
* Accuracy Curve
* Loss Curve

# 9. Histogram

Shows distribution of numerical data.

```python
salary = np.random.normal(
    50000,
    10000,
    1000
)

plt.hist(
    salary,
    bins=20
)

plt.show()
```

Use Cases

* Salary Distribution
* Age Distribution
* Data Normality

---

# 10. Pie Chart

Shows percentage contribution.

```python
sizes = [40,30,20,10]

labels = [
    "Python",
    "Java",
    "C++",
    "JavaScript"
]

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%"
)

plt.show()
```

Use Cases

* Market Share
* Budget Distribution
* Population Distribution

---

# 11. Box Plot

```

* Outlier Detection

# 12. Subplots

plt.show()
```

Useful for comparing multiple graphs.

# 18. Plot Customization

```python

    x,
    y,
    color="red",
    linewidth=3,
    linestyle="--",
    marker="o"

plt.ylabel("Revenue")

plt.show()
```

# 20. Common Interview Questions

## What is Matplotlib?

Matplotlib is a Python library used for creating static, animated, and interactive visualizations.

---

## Difference Between Plot and Scatter

| Plot                 | Scatter              |
| -------------------- | -------------------- |
| Shows trends         | Shows relationships  |
| Connected points     | Individual points    |
| Used for time series | Used for correlation |

---

## Difference Between Histogram and Bar Chart

| Histogram             | Bar Chart        |
| --------------------- | ---------------- |
| Continuous data       | Categorical data |
| Bars touch each other | Bars have gaps   |
| Distribution          | Comparison       |

---

## Why Use Box Plot?

* Detect Outliers
* Median
* Quartiles
* Data Spread

---

## What is Figure Size?

Controls the width and height of the graph.

```python
plt.figure(figsize=(10,6))
```

---

## How to Save a Plot?

```python
plt.savefig(
    "graph.png",
    dpi=300
)
```

---

## Difference Between `plt.plot()` and `plt.bar()`

* `plot()` → Trend Analysis
* `bar()` → Category Comparison

# Best Practices

* Choose the correct chart type for your data.
* Add meaningful titles and axis labels.
* Use legends when plotting multiple datasets.
* Keep visualizations simple and readable.
* Use high-resolution (`dpi=300`) images for reports.
* Combine Matplotlib with Pandas and Seaborn for effective EDA and AI/ML workflows.

# 🎯 Why is Data Visualization Important?

Data visualization helps convert raw data into meaningful insights, making it easier to understand patterns, trends, and relationships before building machine learning models.

Visualization helps answer important questions such as:

* 📈 Is the data normally distributed?
* 🚨 Are there any outliers?
* ⚖️ Is the dataset balanced or imbalanced?
* 🔗 Are features highly correlated?
* 📉 Is the model overfitting?
* 📊 Is the training performance improving over epochs?
* 🎯 Which features are most important?
* 🔍 Are there any hidden patterns in the data?
* 📦 How are different classes distributed?
* ✅ Is the model learning effectively?
