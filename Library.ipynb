```python
arr = np.array([[1,2,3],
                [4,5,6]])

print(arr.size)#size will return the number of elements in the array

print(arr.shape)#shape will return the shape of the array

print(arr.ndim)#ndim will return the number of dimensions of the array

print(arr.dtype)#dtype will return the data type of the array

print(arr.T)#T will return the transpose of the array -- i hope it will be important in future

print(arr.reshape(2,3))#reshape will return the reshaped array

print(arr.flatten())#flatten will return the flattened array

print(arr.tolist())#tolist will return the list of the array

arr = np.array([[1,2,3],
                [4,5,6]])
```

```python
#This is Z-score normalization, commonly used before training algorithms like Logistic Regression, SVM, and Neural Networks.

X = np.array([10, 20, 30, 40, 50])

X_scaled = (X - np.mean(X)) / np.std(X)

X_scaled
```

```python
weights = np.array([0.5, 1.2, -0.7])
features = np.array([2, 4, 1])

prediction = np.dot(weights, features)

print(prediction)
```
