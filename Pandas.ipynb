```python
#Check Missing Values

df.isnull()
df.isnull().sum()
```

```python
df.notnull().sum()
```

```python
#Total Missing Values
df.isnull().sum().sum()
```

```python
#drop bhi kar sakte ho missing values ko


df.dropna() # puri ki puri row hi delete ho jayegi / df.dropna(axis=0)

df.dropna(axis = 1) # jo bhi nan value hogi uska column hi delete ho jayega

df.dropna(how = "all") # agar puri row ya column hi nan hai to hi delete hoga

df.dropna(how = "any") # agar kisi bhi row ya column me nan value hai to hi delete ho jayega

```

    Fill Missing Values

```python
df.fillna(0) # -- filling missing values with 0

df.fillna(df.mean())

df.fillna(df.median())

df.fillna(df.min())

df.fillna(df.max())

df.fillna(df.quantile(0.5))
```

    #if you want to fill the mean value of a column
